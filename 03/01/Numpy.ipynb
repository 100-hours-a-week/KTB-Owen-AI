```python
data10_1 = {
    '이름': ['홍길동', '김철수', '박영희', '이순신'],
    '부서': ['영업', '영업', '인사', '인사'],
    '급여': [5000, 5500, 4800, 5100]
}
df10_1 = pd.DataFrame(data10_1)

df10_1 = df10_1.groupby('부서')['급여'].sum()
print(df10_1)
```

```python
data10_2 = {
    '이름': ['홍길동', '김철수', '박영희', '이순신', '강감찬', '신사임당'],
    '부서': ['영업', '영업', '인사', '인사', 'IT', 'IT'],
    '급여': [5000, 5500, 4800, 5100, 6000, 6200]
}
df10_2 = pd.DataFrame(data10_2)

df10_2 = df10_2.groupby('부서')['급여'].agg(['sum', 'mean'])
print(df10_2)
```

```python
data10_3 = {
    '이름': ['홍길동', '김철수', '박영희', '이순신', '강감찬', '신사임당'],
    '부서': ['영업', '영업', '인사', '인사', 'IT', 'IT'],
    '급여': [5000, 5500, 4800, 5100, 6000, 6200]
}
df10_3 = pd.DataFrame(data10_3)

df10_3 = df10_3.groupby('부서').filter(lambda x: x['급여'].mean() >= 5000)
print(df10_3)
```

```python
df11_1a = pd.DataFrame({'고객ID': [1, 2, 3], '이름': ['홍길동', '김철수', '이영희']})
df11_1b = pd.DataFrame({'고객ID': [2, 3, 4], '구매액': [10000, 20000, 30000]})

df11_1a = pd.merge(df11_1a, df11_1b, on='고객ID', how='inner')
print(df11_1a)
```

```python
df11_2a = pd.DataFrame({'고객ID': [1, 2, 3], '이름': ['홍길동', '김철수', '이영희']})
df11_2b = pd.DataFrame({'고객ID': [2, 3, 4], '구매액': [15000, 25000, 35000]})

df11_2a = pd.merge(df11_2a, df11_2b, on='고객ID', how='left')
print(df11_2a)
```

```python
df11_3a = pd.DataFrame({
    '고객ID': [1, 2, 3],
    '도시': ['서울', '부산', '대전'],
    '구매액': [10000, 20000, 30000]
})
df11_3b = pd.DataFrame({
    '고객ID': [1, 2, 3],
    '도시': ['서울', '부산', '광주'],
    '구매액': [15000, 25000, 35000]
})

df11_3a = pd.merge(df11_3a, df11_3b, on=['고객ID', '도시'], how='outer', suffixes=('_left', '_right'))
print(df11_3a)
```

```python
data13_1 = {
    '날짜': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250]
}
df13_1 = pd.DataFrame(data13_1)

df13_1 = df13_1.pivot(index='날짜', columns='제품', values='판매량')
print(df13_1)
```

```python
data13_2 = {
    '카테고리': ['전자', '가전', '전자', '가전'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250]
}
df13_2 = pd.DataFrame(data13_2)

df13_2 = df13_2.pivot_table(index='카테고리', columns='제품', values='판매량', aggfunc='sum')
print(df13_2)
```

```python
data13_3 = {
    '날짜': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250],
    '이익': [20, 50, 30, 60]
}
df13_3 = pd.DataFrame(data13_3)

df13_3 = df13_3.pivot(index='날짜', columns='제품', values=['판매량', '이익'])
print(df13_3)
```
