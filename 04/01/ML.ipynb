```python
from sklearn.datasets import make_classification

# 가상 사이킷런 데이터 생성

X, y = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=10,
    n_redundant=5,
    n_classes=2,
    class_sep=1.2,
    flip_y=0.10, # 노이즈 10% 
    random_state=42,
    weights=[0.8, 0.2],
)
print("------------------------")
print(X.shape)
print(X[:5])
print("----------------")
print(y.shape)
print(y[:5])
```

1. 가상 데이터셋을 생성한 뒤, 학습·검증·테스트 데이터셋으로 분할해 보세요.


```python
from sklearn.model_selection import train_test_split
# 1차분할 -> 8:2 train:test
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, stratify=y,random_state=42)

# 2차 분할 (8:1:1) train:test:valid
X_valid, X_test, y_valid, y_test = train_test_split(X_test, y_test, test_size=0.5, stratify=y_test,random_state=42)
```

```python
print("X_train:", X_train.shape)
print("X_valid:", X_valid.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_valid:", y_valid.shape)
print("y_test :", y_test.shape)

print("-------------------------------")

from collections import Counter
# stratify 확인

print("y_train:", Counter(y_train))
print("y_valid:", Counter(y_valid))
print("y_test :", Counter(y_test))
```

2. 가상 데이터셋을 생성하고, K-최근접이웃(K-NN) 알고리즘으로 학습·예측을 수행해 보세요.


- KNN
    - 가장 가까운 거리 

```python
from sklearn.metrics import accuracy_score,classification_report
from sklearn.preprocessing import StandardScaler

# scaler로 스케일 먼저. 
scaler = StandardScaler()

X_train_scale = scaler.fit_transform(X_train) # fit까지 합쳐서 하는 애임 
X_valid_scale = scaler.transform(X_valid)
X_test_scale = scaler.transform(X_test)

from sklearn.neighbors import KNeighborsClassifier

# 모델 정의 
knn = KNeighborsClassifier(
    n_neighbors=5
)
# 모델 학습 fit 
knn.fit(X_train_scale, y_train)
# 모델 예측 
y_pred = knn.predict(X_test_scale)

print("Accuracy : " , accuracy_score(y_test,y_pred))
print("classification_report : " , classification_report(y_test,y_pred))
```

3. 동일한 이진 분류 가상 데이터셋을 생성하고, Perceptron, SVM, Random Forest, Naive Bayes 네 가지 알고리즘으로 학습해 보세요.


- perceptron
    - 나누는 직선 하나 찾기 

```python
# perceptron
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score, classification_report
```

```python
perceptron = Perceptron(random_state=42)

perceptron.fit(X_train_scale, y_train)

y_pred = perceptron.predict(X_test_scale)

print("perceptron Accuracy: ",accuracy_score(y_test,y_pred))
print("classification_report",classification_report(y_test,y_pred))
```

- SVM
    - 나눠지기만 하는게 아니라 안전하게 나누는 직선 찾기 

```python
from sklearn.svm import SVC # LinearSVC, SVR 써도됨 
```

```python
svm = SVC(kernel="rbf", random_state=42)

svm.fit(X_train_scale,y_train)

y_pred = svm.predict(X_test_scale)

print("SVM Accuracy:", accuracy_score(y_test,y_pred))

print("classification_report", classification_report(y_test, y_pred))
```

- Random Forest
    - 앙상블로 분류 

```python
from sklearn.ensemble import RandomForestClassifier

random_forest = RandomForestClassifier(n_estimators=100, random_state=42)

random_forest.fit(X_train_scale,y_train)

y_pred = random_forest.predict(X_test_scale)

print("random_forest Accuracy:", accuracy_score(y_test,y_pred))

print("classification_report", classification_report(y_test, y_pred))
```

- Gaussian Naive Bayes
    - 확률 기반 분류 
    

```python
from sklearn.naive_bayes import GaussianNB 

gaussian_nb = GaussianNB()
gaussian_nb.fit(X_train_scale,y_train)

y_pred = gaussian_nb.predict(X_test_scale)

print("gaussian_nb Accuracy:", accuracy_score(y_test,y_pred))

print("classification_report", classification_report(y_test, y_pred))
```

4.  가상 데이터셋을 준비하고, 증강(Data Augmentation) 기법을 적용했을 때와 적용하지 않았을 때 모델 성능을 비교하세요.
- "uv add imbalanced-learn" 사용 

```python
svm = SVC(kernel="rbf", random_state=42)

svm.fit(X_train_scale,y_train)

y_pred = svm.predict(X_test_scale)

print("증강전 ")

print("SVM Accuracy:", accuracy_score(y_test,y_pred))
print("classification_report", classification_report(y_test, y_pred))

from imblearn.over_sampling import SMOTE
print("증강후 ")

smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train_scale,y_train) 

svm_smote = SVC(kernel="rbf", random_state=42)
svm_smote.fit(X_train_smote, y_train_smote)
y_pred = svm_smote.predict(X_test_scale)

print("SVM Accuracy:", accuracy_score(y_test,y_pred))
print("classification_report", classification_report(y_test, y_pred))
```
